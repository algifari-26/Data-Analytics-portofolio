```python
import pandas as pd
```

```python
df = pd.read_csv('/train.csv')
```

```python
df.head()
```

```python
print("\n--- Ringkasan Dataset ---")

df.info()
```

```python
# Menampilkan baris yang kolom 'Postal Code'-nya kosong (NaN)
df_kosong = df[df['Postal Code'].isnull()]
display(df_kosong[['Order ID', 'Customer Name', 'City', 'State', 'Postal Code']])
```

```python
# Mengisi (Fill) nilai kosong pada Postal Code dengan angka 5401
df['Postal Code'] = df['Postal Code'].fillna(5401)

# Mengubah tipe data Postal Code dari float (desimal) menjadi string (teks)
# Karena kodepos bukan angka untuk dijumlahkan/dikurangi
df['Postal Code'] = df['Postal Code'].astype(int).astype(str)
```

```python
# Mengubah format data dan Menambahkan dayfirst=True agar Pandas tahu formatnya Tanggal/Bulan/Tahun
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)
```

```python
print("\n--- Ringkasan Dataset ---")

df.info()
```

```python
# Menyimpan dataset yang sudah bersih menjadi file CSV baru
df.to_csv('Superstore_Cleaned.csv', index=False)
print("Data bersih berhasil disimpan sebagai 'Superstore_Cleaned.csv'!")
```
